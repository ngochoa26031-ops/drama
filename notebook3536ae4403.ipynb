```python
import subprocess, sys
from pathlib import Path

ROOT = Path('/kaggle/working')
LOADER_REPO = ROOT / 'keepsfx_loader'

subprocess.run(['rm', '-rf', str(LOADER_REPO)], check=False)
subprocess.run(['git', 'clone', '-q', 'https://github.com/yaylbanh/keepsfx.git', str(LOADER_REPO)], check=True)
subprocess.run(['git', '-C', str(LOADER_REPO), 'checkout', '-q', 'kaggle'], check=True)
subprocess.run([sys.executable, str(LOADER_REPO / 'run_kaggle.py')], check=True)
```
